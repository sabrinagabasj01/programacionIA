## 0. ¿Qué es Tkinter?

**Tkinter** es la librería estándar de Python para crear interfaces gráficas (**GUI**).  
Ideas clave:
- Un programa Tkinter crea una **ventana raíz** (`Tk`), añade **widgets** (etiquetas, botones, entradas, etc.) y arranca un **bucle de eventos** (`mainloop()`).
- Los **gestores de geometría** (`pack`, `grid`, `place`) controlan la posición de los widgets.  
- En nuestro proyecto de chatbot usaremos: `Label`, `Entry`, `Button`, `ScrolledText`, `messagebox`/`simpledialog`, y manejo de eventos.

> Este cuaderno solo enseña lo **necesario** para nuestro objetivo final.

## 1. Ventana y etiqueta
Crea una ventana con un título y una etiqueta de texto.

````python
import tkinter as tk

def ventana_basica():
    """
        Creamos una ventana raíz tk, le ponemos un titulo y mostramos una etiqueta de texto.
    """
    
    # 1) Creamos la ventana 
    root = tk.Tk() # Inicializa la ventana principal (root)
    
    # 2) Cambiamos el objeto de la ventana
    root.title("Mi primera ventana")
    
    # 3) Le damos una etiqueta de texto
    
    etiqueta = tk.Label(root, text= "Hola PIA!")
    
    # 4) Mostrar etiqueta 
    etiqueta.pack(padx= 20, pady= 50)
    
    # x) Inicializamos el bucle que muestra la ventana 
    root.mainloop()
    
ventana_basica()
````

## 2. Inputs y botones 
Añadimos un `Entry` para escribir un nombre y un `Button` que saluda al pulsarse.

````python
import tkinter as tk
from tkinter import messagebox # Importar el paquete para añadir cajas de mensaje

def app_saludo():
    """
        Crearemos una ventana con un Entry (input)  un boton 
        Cuando se pulse el boton, se muestra un saludo en un cuadro de dialogo
    """ 
    
    # 1) Creamos objeto ventana 
    root = tk.Tk()
    
    # 2) Añadimos un titulo
    root.title("EL saludador")
    
    # 3) Añadimos una etiqueta explicativa
    tk.Label(root, text= "Escribe tu nombre: ").pack(padx=10, pady=(10, 0))
    
    # 4) Introduciomos un campo de entrada 
    entrada = tk.Entry(root) # Entry nos permite escribir texto
    entrada.pack(padx=20, pady=5)
    
    # 5) Funcion que se ejecuta cuando pulsamos un boton
    def saludar():
        # Saludar utilizara el metodo .get() para recuperar el texto de la entrada
        nombre = entrada.get().strip().capitalize()
        # Para evitar problemas si no escribe nada el usuario, le indicaremos que el input esta vacio
        if not nombre:
            messagebox.showwarning("Aviso", "Por favor di tu nombre.")
            return
        messagebox.showinfo("Saludos:", f"{nombre}")
        
    # 6) Creamos un boton que dispara la funcion 
    tk.Button(root, text="Saludar", command=saludar).pack(pady=10)
    
    # X) Bucle principal de la ventana
    root.mainloop()
    
app_saludo()
````

## 3. Layout con `grid` (ordenar la interfaz)
`grid` organiza los widgets **por filas y columnas**. Haremos una mini-tarea: concatenar nombre y apellido.

````python
import tkinter as tk

def app_grid():
    """
        Interfaz con dos Entry (nombre y apellido) y Label con el resultado concatenado
    """
    root = tk.Tk()
    root.title("Grid: concatenar")
    
    # Fila 0: Etiqueta y entrada para el nombre 
    tk.Label(root, text="Nombre: ").grid(row=0, column=0, padx=10, pady=10, sticky="e")
    entrada_nombre = tk.Entry(root, width=20)
    entrada_nombre.grid(row=0, column=1, padx=10, pady=10, sticky="e")
    
    # Fila 1: Etiqueta y entrada para apellido
    tk.Label(root, text="Apellido: ").grid(row=1, column=0, padx=10, pady=10, sticky="e")
    entrada_apellido = tk.Entry(root, width=20)
    entrada_apellido.grid(row=1, column=1, padx=10, pady=10, sticky="e")
    
    # Fila 2:  Boton y resultado
    resultado = tk.Label(root, text="El resultado es")
    resultado.grid(row=2, column=0, columnspan=2, padx=10, pady=10 , sticky="w")
    
    # Función que concatena los valores
    def concatenar():
        nombre = entrada_nombre.get().strip()
        apellido = entrada_apellido.get().strip()
        resultado.config(text=f"resultado: {nombre} {apellido}".strip())
        
    # Botón para concatenar
    tk.Button(root, text="Concatenar", command=concatenar).grid(row=2, column=1, columnspan=2, pady=10, padx=(100, 20))
    
    # Mainloop
    root.mainloop()
    
app_grid()
````

## 4. Variables de control (`StringVar`) y actualización en vivo
`StringVar` mantiene un valor y **notifica cambios**. Actualizaremos un contador de caracteres mientras escribes.

````python
import tkinter as tk

def app_contador_caracteres():
    """
        Mostrar un Entry y una etiqueta que cuente caracteres en tiempo real 
    """

    root = tk.Tk()
    root.title("Contador de caracteres.")
    
    # Creamos una variable de tipo stringVar que almacenara el texto
    var_texto = tk.StringVar(value="")
    
    # Creamos un Entry 
    tk.Entry(root, textvariable=var_texto, width=50). pack(padx=10, pady=10)
    
    # Creamos la etiqueta 
    etiqueta = tk.Label(root, text="0 caracteres")
    etiqueta.pack(padx=10, pady=10)
    
    # Nuestro objetivo es que suceda algo cuando un elemento le afecte un evento
    # 1. Definimos el "algo" --> funcion
    
    def cambiar_text(*args): # No tenemos claro cuantos argumentos de entrada en la funcion y utilizaremos args
        longitud = len(var_texto.get())
        if longitud ==1:
            etiqueta.config(text=f"{longitud} caracter.")
            return
        
        etiqueta.config(text=f"{longitud} caracter.")
        
    # Añadir evento de escucha al escribir
    var_texto.trace_add("write", cambiar_text)
        
    # Mainloop
    root.mainloop()
    
app_contador_caracteres()
````

## 11. Chatbot con **diccionario de patrones**
Evolucionamos a un **diccionario** que mapea patrones (claves) a **listas de respuestas**.
Este esquema facilita **añadir**/modificar respuestas.

## 12. Integración del chatbot (diccionario) en Tkinter
Unimos lo aprendido: `ScrolledText` para el historial, `Entry` para escribir, `Button` y **Enter** para enviar.

````python

import tkinter as tk
from tkinter.scrolledtext import ScrolledText
import random

PATRONES = {
    "hola": ["¡Hola! ¿En qué puedo ayudarte?", "¡Buenas! ¿Qué necesitas?"],
    "adios": ["¡Hasta luego!", "¡Adiós! Que tengas buen día."],
    "como estas": ["¡Muy bien! ¿Y tú?", "Aquí, listo para ayudarte."]
}

def obtener_respuesta_diccionario(mensaje):
    """
    Igual que antes: búsqueda exacta y por inclusión.
    """
    msg = mensaje.lower().strip()
    if msg in PATRONES:
        return random.choice(PATRONES[msg])
    for patron, respuestas in PATRONES.items():
        if patron in msg:
            return random.choice(respuestas)
    return "No tengo respuesta para eso todavía."

def app_chatbot_diccionario():
    """
    Interfaz mínima de chatbot basada en diccionario de patrones.
    """
    root = tk.Tk()
    root.title("ChatBot (diccionario)")

    chat = ScrolledText(root, wrap=tk.WORD, width=50, height=15)
    chat.pack(padx=10, pady=10)

    entrada = tk.Entry(root, width=50)
    entrada.pack(padx=10, pady=5)
    entrada.focus_set()

    def enviar(event=None):
        texto = entrada.get().strip()
        if not texto:
            return
        chat.insert(tk.END, f"Tú: {texto}\n")
        entrada.delete(0, tk.END)

        respuesta = obtener_respuesta_diccionario(texto)
        chat.insert(tk.END, f"Bot: {respuesta}\n")
        chat.see(tk.END)

    tk.Button(root, text="Enviar", command=enviar).pack(pady=5)
    root.bind("<Return>", enviar)

    root.mainloop()

# Ejecuta la demo:
app_chatbot_diccionario()
````

## 13. Formato **JSON** de intents (teoría)

Usaremos un **archivo JSON** para guardar **intents**: cada intent tiene `patterns` (lista de frases que el bot reconoce) y `responses` (lista de posibles respuestas).
Estructura típica:

```json
{
  "intents": [
    {
      "patterns": ["Hola", "Buenas", "Como estas?"],
      "responses": ["Hola! En que puedo ayudarte?", "Buenas! Que necesitas?"]
    },
    {
      "patterns": ["Adios", "Hasta luego"],
      "responses": ["Adios! Que tengas un buen dia.", "Hasta luego, cuidate!"]
    }
  ]
}
```

Ventajas:
- Añadir conocimientos sin tocar el código.
- Posibilidad de **aprender** nuevos intents durante la ejecución y **guardar** los cambios.
