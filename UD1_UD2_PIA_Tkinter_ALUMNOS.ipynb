## 0. ¿Qué es Tkinter?

**Tkinter** es la librería estándar de Python para crear interfaces gráficas (**GUI**).  
Ideas clave:
- Un programa Tkinter crea una **ventana raíz** (`Tk`), añade **widgets** (etiquetas, botones, entradas, etc.) y arranca un **bucle de eventos** (`mainloop()`).
- Los **gestores de geometría** (`pack`, `grid`, `place`) controlan la posición de los widgets.  
- En nuestro proyecto de chatbot usaremos: `Label`, `Entry`, `Button`, `ScrolledText`, `messagebox`/`simpledialog`, y manejo de eventos.

> Este cuaderno solo enseña lo **necesario** para nuestro objetivo final.

## 1. Ventana y etiqueta
Crea una ventana con un título y una etiqueta de texto.

````python
import tkinter as tk

def ventana_basica():
    """
        Creamos una ventana raíz tk, le ponemos un titulo y mostramos una etiqueta de texto.
    """
    
    # 1) Creamos la ventana 
    root = tk.Tk() # Inicializa la ventana principal (root)
    
    # 2) Cambiamos el objeto de la ventana
    root.title("Mi primera ventana")
    
    # 3) Le damos una etiqueta de texto
    
    etiqueta = tk.Label(root, text= "Hola PIA!")
    
    # 4) Mostrar etiqueta 
    etiqueta.pack(padx= 20, pady= 50)
    
    # x) Inicializamos el bucle que muestra la ventana 
    root.mainloop()
    
ventana_basica()
````

## 2. Inputs y botones 
Añadimos un `Entry` para escribir un nombre y un `Button` que saluda al pulsarse.

````python
import tkinter as tk
from tkinter import messagebox # Importar el paquete para añadir cajas de mensaje

def app_saludo():
    """
        Crearemos una ventana con un Entry (input)  un boton 
        Cuando se pulse el boton, se muestra un saludo en un cuadro de dialogo
    """ 
    
    # 1) Creamos objeto ventana 
    root = tk.Tk()
    
    # 2) Añadimos un titulo
    root.title("EL saludador")
    
    # 3) Añadimos una etiqueta explicativa
    tk.Label(root, text= "Escribe tu nombre: ").pack(padx=10, pady=(10, 0))
    
    # 4) Introduciomos un campo de entrada 
    entrada = tk.Entry(root) # Entry nos permite escribir texto
    entrada.pack(padx=20, pady=5)
    
    # 5) Funcion que se ejecuta cuando pulsamos un boton
    def saludar():
        # Saludar utilizara el metodo .get() para recuperar el texto de la entrada
        nombre = entrada.get().strip().capitalize()
        # Para evitar problemas si no escribe nada el usuario, le indicaremos que el input esta vacio
        if not nombre:
            messagebox.showwarning("Aviso", "Por favor di tu nombre.")
            return
        messagebox.showinfo("Saludos:", f"{nombre}")
        
    # 6) Creamos un boton que dispara la funcion 
    tk.Button(root, text="Saludar", command=saludar).pack(pady=10)
    
    # X) Bucle principal de la ventana
    root.mainloop()
    
app_saludo()
````

## 3. Layout con `grid` (ordenar la interfaz)
`grid` organiza los widgets **por filas y columnas**. Haremos una mini-tarea: concatenar nombre y apellido.

````python
import tkinter as tk

def app_grid():
    """
        Interfaz con dos Entry (nombre y apellido) y Label con el resultado concatenado
    """
    root = tk.Tk()
    root.title("Grid: concatenar")
    
    # Fila 0: Etiqueta y entrada para el nombre 
    tk.Label(root, text="Nombre: ").grid(row=0, column=0, padx=10, pady=10, sticky="e")
    entrada_nombre = tk.Entry(root, width=20)
    entrada_nombre.grid(row=0, column=1, padx=10, pady=10, sticky="e")
    
    # Fila 1: Etiqueta y entrada para apellido
    tk.Label(root, text="Apellido: ").grid(row=1, column=0, padx=10, pady=10, sticky="e")
    entrada_apellido = tk.Entry(root, width=20)
    entrada_apellido.grid(row=1, column=1, padx=10, pady=10, sticky="e")
    
    # Fila 2:  Boton y resultado
    resultado = tk.Label(root, text="El resultado es")
    resultado.grid(row=2, column=0, columnspan=2, padx=10, pady=10 , sticky="w")
    
    # Función que concatena los valores
    def concatenar():
        nombre = entrada_nombre.get().strip()
        apellido = entrada_apellido.get().strip()
        resultado.config(text=f"resultado: {nombre} {apellido}".strip())
        
    # Botón para concatenar
    tk.Button(root, text="Concatenar", command=concatenar).grid(row=2, column=1, columnspan=2, pady=10, padx=(100, 20))
    
    # Mainloop
    root.mainloop()
    
app_grid()
````

## 4. Variables de control (`StringVar`) y actualización en vivo
`StringVar` mantiene un valor y **notifica cambios**. Actualizaremos un contador de caracteres mientras escribes.

````python
import tkinter as tk

def app_contador_caracteres():
    """
        Mostrar un Entry y una etiqueta que cuente caracteres en tiempo real 
    """

    root = tk.Tk()
    root.title("Contador de caracteres.")
    
    # Creamos una variable de tipo stringVar que almacenara el texto
    var_texto = tk.StringVar(value="")
    
    # Creamos un Entry 
    tk.Entry(root, textvariable=var_texto, width=50). pack(padx=10, pady=10)
    
    # Creamos la etiqueta 
    etiqueta = tk.Label(root, text="0 caracteres")
    etiqueta.pack(padx=10, pady=10)
    
    # Nuestro objetivo es que suceda algo cuando un elemento le afecte un evento
    # 1. Definimos el "algo" --> funcion
    
    def cambiar_text(*args): # No tenemos claro cuantos argumentos de entrada en la funcion y utilizaremos args
        longitud = len(var_texto.get())
        if longitud ==1:
            etiqueta.config(text=f"{longitud} caracter.")
            return
        
        etiqueta.config(text=f"{longitud} caracter.")
        
    # Añadir evento de escucha al escribir
    var_texto.trace_add("write", cambiar_text)
        
    # Mainloop
    root.mainloop()
    
app_contador_caracteres()
````

## 5. Eventos y `bind` (tecla Enter)
Podemos **enlazar eventos** del teclado/ratón a funciones. Aquí: pulsar **Enter** equivale a pulsar el botón.

````python
import tkinter as tk

def bind_enter():
    """
        Vamos a usar .bind() para asociar la tecla enter a una funcion
    """
    
    root = tk.Tk()
    root.title("Bind de Enter")
    
    entrada = tk.Entry(root, width=50)
    entrada.pack(padx=10, pady=10)
    entrada.focus_set() # Colocamos el cursor en esta entrada
    
    salida = tk.Label(root, text="Pulsa enter para continuar.")
    salida.pack(padx=10, pady=10)
    
    # Declaramos una funcion que se ejecutara al pulsar enter
    def confirmar(event=None):
        # Event es opcional y captura la informacion de un evento
        # como presionar una tecla. Event a none permite varios eventos 
        # puedan llamar a la misma funcion  
        
        salida.config(text=f"Confirmado: {entrada.get()}.")  
    
    root.bind("<Return>", confirmar)
        
    
    
    root.mainloop()
    
bind_enter()
````

## 6. `Text` / `ScrolledText` (área de conversación)
`ScrolledText` añade una **barra de scroll** automáticamente. Útil para el historial del chat.

````python
import tkinter as tk
from tkinter.scrolledtext import ScrolledText

def app_scrolled():
    
    """
        Crear un area de texto con scroll para imprimir mensajes
    """
    
    root = tk.Tk()
    root.title("Scrolled Text")
    
    # Creamos el area de tecto con scroll
    chat = ScrolledText(root, wrap=tk.WORD, width= 50, height=10, state=tk.DISABLED) # tk.WORD configura el salto de linea estre palabras
    chat.pack(padx=10, pady=10)
    
    # Creamos la entrada 
    entrada = tk.Entry(root, width=50)
    entrada.pack(padx=10, pady=10)
    entrada.focus_set()
    
    def enviar(event=None):
        
        chat.config(state=tk.NORMAL)
        
        texto = entrada.get().strip()
        
        # Evitamos textos vacios
        if not texto:
            return
                
        chat.insert(tk.END, f"[USUARIO] {texto}\n")
        
        # Desplazar automaticamente el scroll al final
        chat.see(tk.END)
        
        # Limpiamos el campo de entrada
        entrada.delete(0, tk.END) # Delete borra el contenido desde x hasta 
        
        chat.config(state=tk.DISABLED)
        
    root.bind("<Return>", enviar)
    
    root.mainloop()

app_scrolled()
    
````

## 8. Imágenes con `PhotoImage` (manejo seguro)
`PhotoImage` carga imágenes (PNG/GIF). Usaremos un `try/except` por si el archivo no existe.

````python
import tkinter as tk
from pathlib import Path

def app_imagen():
    """
        Intenta cargar una imagen de un directorio y si no existe muestra un texto
        alternativo
    """  

    root = tk.Tk()
    root.title("Imagen")
    
    ruta = Path("img/imagen2.png")
    
    # Contenedor de la imagen
    IMG_WIDTH = 100
    IMG_HEIGHT = 100
    
    etiqueta = tk.Label(root, text="Aqui deberia verse una imagen")
    etiqueta.pack(padx=10, pady=10)
    
    try: 
        if ruta.exists():
            img = tk.PhotoImage(file=str(ruta))
            # img = img.subsample(int(img.width()/IMG_WIDTH), int(img.height()/IMG_HEIGHT))
                        
            etiqueta.config(image=img, text="")
        else:
            etiqueta.config(text="No se encontro la imagen.")
    except Exception as e: 
        etiqueta.config(text=f"No se pudo cargar la imagen. Error{e}")
    
    root.mainloop()
        
app_imagen()
````

## 9. Organización con `Frame` y opciones `expand/fill`
`Frame` agrupa widgets. `pack(expand=True, fill='both')` permite que se **adapten** al tamaño de la ventana.

````python
import tkinter as tk

def app_frame():
    root = tk.Tk()
    root.title("Frame")
    
    frame_top = tk.Frame(root, bg="#6D44FF")
    frame_bottom = tk.Frame(root, bg="#ff0008")
    
    # expand -> nos permite ocupar todo el espacio disponible de la ventana 
    # fill = "both", rellena el ancho y el alto
    
    frame_top.pack(expand=True, fill="both")
    frame_bottom.pack(expand=True, fill="x")
    
    # Creamos dos etiquetas pero indicamos que se encuentran en los frames
    tk.Label(frame_top, text="Parte superior").pack(pady=20)
    tk.Label(frame_bottom, text="Parte superior").pack(pady=20)
    
    root.mainloop()



app_frame()
````

## 10. Estructura básica de un chatbot (con **listas**)
Primero, sin interfaz gráfica. Detectamos respuestas comparando el texto con **listas** de patrones.
- Si coincide exactamente con un patrón conocido, respondemos.
- Si no, intentamos coincidencias **simples** (por ejemplo, que contenga la palabra clave).

## 11. Chatbot con **diccionario de patrones**
Evolucionamos a un **diccionario** que mapea patrones (claves) a **listas de respuestas**.
Este esquema facilita **añadir**/modificar respuestas.

````python
import random

PATRONES = {
    # Cada clave es un 'patrón' textual sencillo
    "hola": ["¡Hola! ¿En qué puedo ayudarte?", "¡Buenas! ¿Qué necesitas?"],
    "adios": ["¡Hasta luego!", "¡Adiós! Que tengas buen día."],
    "como estas": ["¡Muy bien! ¿Y tú?", "Aquí, listo para ayudarte."]
}

def obtener_respuesta(mensaje):
    """
        Buscar una respuesta aleatoria en el diccionario PATRONES
    """
    
    mensaje = mensaje.lower().strip()
    
    letras_con = ["á", "é", "í", "ó", "ú"]
    letras_sin = ["a", "e", "i", "o", "u"]
    
    
    # Por coincidencias exactas 
    
    mensaje = list(mensaje)
    
    for i, letra in enumerate (mensaje):
        for j, con in enumerate(letras_con):
            if letra == con:
                mensaje[i] = letras_sin[j]
    
    mensaje = "".join(mensaje)
    # .unicode(s) --> cambia el abecedario en inglés
    
    if mensaje in PATRONES.items():
        return random.choice(PATRONES[mensaje])   
    
    # Por inclusion de contenido
    
    for patron, respuestas in PATRONES.items():
        if patron in mensaje:
            return random.choice(respuestas)  
        
    return "No tengo respuesta para eso"       
    
# Pruebas rápidas
frases = ["hola", "¿Cómo estás?", "Bueno, adiós"]

for prueba in frases:
    print("[USUARIO] ", prueba)
    print("[ChatBot] ", obtener_respuesta(prueba))
````

## 12. Integración del chatbot (diccionario) en Tkinter
Unimos lo aprendido: `ScrolledText` para el historial, `Entry` para escribir, `Button` y **Enter** para enviar.

````python

import tkinter as tk
from tkinter.scrolledtext import ScrolledText
import random

PATRONES = {
    "hola": ["¡Hola! ¿En qué puedo ayudarte?", "¡Buenas! ¿Qué necesitas?"],
    "adios": ["¡Hasta luego!", "¡Adiós! Que tengas buen día."],
    "como estas": ["¡Muy bien! ¿Y tú?", "Aquí, listo para ayudarte."]
}

def obtener_respuesta_diccionario(mensaje):
    """
    Igual que antes: búsqueda exacta y por inclusión.
    """
    msg = mensaje.lower().strip()
    if msg in PATRONES:
        return random.choice(PATRONES[msg])
    for patron, respuestas in PATRONES.items():
        if patron in msg:
            return random.choice(respuestas)
    return "No tengo respuesta para eso todavía."

def app_chatbot_diccionario():
    """
    Interfaz mínima de chatbot basada en diccionario de patrones.
    """
    root = tk.Tk()
    root.title("ChatBot (diccionario)")

    chat = ScrolledText(root, wrap=tk.WORD, width=50, height=15)
    chat.pack(padx=10, pady=10)

    entrada = tk.Entry(root, width=50)
    entrada.pack(padx=10, pady=5)
    entrada.focus_set()

    def enviar(event=None):
        texto = entrada.get().strip()
        if not texto:
            return
        chat.insert(tk.END, f"Tú: {texto}\n")
        entrada.delete(0, tk.END)

        respuesta = obtener_respuesta_diccionario(texto)
        chat.insert(tk.END, f"Bot: {respuesta}\n")
        chat.see(tk.END)

    tk.Button(root, text="Enviar", command=enviar).pack(pady=5)
    root.bind("<Return>", enviar)

    root.mainloop()

# Ejecuta la demo:
app_chatbot_diccionario()
````

## 13. Formato **JSON** de intents (teoría)

Usaremos un **archivo JSON** para guardar **intents**: cada intent tiene `patterns` (lista de frases que el bot reconoce) y `responses` (lista de posibles respuestas).
Estructura típica:

```json
{
  "intents": [
    {
      "patterns": ["Hola", "Buenas", "Como estas?"],
      "responses": ["Hola! En que puedo ayudarte?", "Buenas! Que necesitas?"]
    },
    {
      "patterns": ["Adios", "Hasta luego"],
      "responses": ["Adios! Que tengas un buen dia.", "Hasta luego, cuidate!"]
    }
  ]
}
```

Ventajas:
- Añadir conocimientos sin tocar el código.
- Posibilidad de **aprender** nuevos intents durante la ejecución y **guardar** los cambios.

## 14. Cargar intents desde JSON y responder
Leemos el JSON, extraemos todos los **patrones** y sus **respuestas** y buscamos coincidencias.
Usamos una coincidencia sencilla (exacta o contención). Si quieres, puedes mejorarla con `difflib.get_close_matches`.

## 18. Difflib: Comparación de secuencias
 
La librería estándar `difflib` es muy útil cuando trabajamos con texto y queremos encontrar similitudes o diferencias entre cadenas. En el contexto de un chatbot, es fundamental para manejar **errores tipográficos** o entradas que no coinciden exactamente con lo que esperamos.
 
### Funciones clave:
 
1.  **`SequenceMatcher(None, a, b).ratio()`**:
    *   Compara dos secuencias `a` y `b`.
    *   Devuelve un valor flotante entre `0.0` (completamente diferentes) y `1.0` (idénticas).
    *   Es útil para saber "cuánto" se parecen dos frases.
 
2.  **`get_close_matches(word, possibilities, n=3, cutoff=0.6)`**:
    *   Busca las mejores coincidencias de `word` dentro de una lista `possibilities`.
    *   `n`: número máximo de resultados a devolver.
    *   `cutoff`: puntuación mínima de similitud (0 a 1) para considerar una coincidencia válida.
    *   Devuelve una lista con las coincidencias encontradas, ordenadas por similitud.

````python
import difflib

def Chatbot_Diff():
    """
    Simula un chatbot básico que intenta entender frases usando difflib.
    """

    # Frases: base de conocimiento
    frases_conocidas = [
        "hola",
        "enciende la luz",
        "abre la puerta"
    ]

    # Respuestas asociadas
    respuestas = {
        "hola": "¡Hola! Soy yo, Wheatley. ¿Me recibes?",
        "enciende la luz": "¡Hágase la luz!... Ah, espera, no tengo manos.",
        "abre la puerta": "Intentando hackear la puerta... ¡Ajá! No, espera, eso era la luz."
    }

    print("--- WHEATLEY OS v1.0 ---")
    print("Intenta hablar con Wheatley (escribe 'salir' para terminar)")
    print("Frases que conoce: 'hola', 'enciende la luz', 'abre la puerta'")
    print()

    while True:
        usuario = input("Tú: ").strip().lower()

        if usuario == "salir":
            print("Wheatley: Cerrando sesión... ¡no me olvides! 👋")
            break

        if not usuario:
            print("Wheatley: ¿Silencio incómodo? Prueba a decirme algo.")
            continue

        # Buscar la frase más parecida usando get_close_matches
        coincidencias = difflib.get_close_matches(
            usuario,
            frases_conocidas
            n=1,       # solo queremos la mejor
            cutoff=0.4 # mínimo de similitud (0 a 1)
        )

        if coincidencias:
            mejor_frase = coincidencias[0]

            # Calcular ratio de similitud para mostrarlo (informativo)
            ratio = difflib.SequenceMatcher(None, usuario, mejor_frase).ratio()
            porcentaje = round(ratio * 100, 1)

            print(f"(He entendido: '{mejor_frase}' - similitud: {porcentaje}%)")
            print("Wheatley:", respuestas[mejor_frase])
        else:
            print("Wheatley: No tengo ni idea de lo que quieres decir. 🤯")
            print("          (Prueba algo como 'hola', 'enciende la luz' o 'abre la puerta')")

if __name__ == "__main__":
    Chatbot_Diff()
````
